```python
import pandas as pd
import numpy as np

# Create student data
data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Anu", "Reethu", "Soni", "Keerthi", "Sai"],
    "Python": [85, 72, 90, 65, 78],
    "Maths": [80, 68, 92, 70, 75],
    "English": [75, 80, 88, 60, 70],
    "Attendance": [92, 85, 96, 75, 82]
}

# Convert data into Pandas DataFrame
df = pd.DataFrame(data)

# Calculate total marks using Pandas
subjects = ["Python", "Maths", "English"]

df["Total"] = df[subjects].sum(axis=1)

# Calculate average using NumPy
df["Average"] = np.mean(df[subjects], axis=1)

# Calculate percentage
df["Percentage"] = (df["Total"] / 300) * 100

# Assign performance level
def get_performance(mark):
    if mark >= 90:
        return "Excellent"
    elif mark >= 75:
        return "Good"
    elif mark >= 60:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Percentage"].apply(get_performance)

# Display complete student report
print("\n===== STUDENT PERFORMANCE REPORT =====")
print(df.to_string(index=False))

# Class statistics using NumPy
print("\n===== CLASS ANALYSIS =====")

print("Class Average:",
      round(np.mean(df["Percentage"]), 2), "%")

print("Highest Percentage:",
      round(np.max(df["Percentage"]), 2), "%")

print("Lowest Percentage:",
      round(np.min(df["Percentage"]), 2), "%")

# Find top student
top_index = np.argmax(df["Percentage"])

print("\nTop Performing Student:")
print("Name:", df.loc[top_index, "Name"])
print("Percentage:",
      round(df.loc[top_index, "Percentage"], 2), "%")

# Subject-wise average
print("\n===== SUBJECT-WISE AVERAGE =====")

for subject in subjects:
    average = np.mean(df[subject])
    print(subject, ":", round(average, 2))

# Attendance analysis
print("\n===== ATTENDANCE ANALYSIS =====")

low_attendance = df[df["Attendance"] < 75]

if len(low_attendance) > 0:
    print("Students with attendance below 75%:")
    print(low_attendance[["Name", "Attendance"]].to_string(index=False))
else:
    print("All students have attendance above 75%.")

# Save report
df.to_csv("student_performance_report.csv", index=False)

print("\nReport saved successfully!")
```
