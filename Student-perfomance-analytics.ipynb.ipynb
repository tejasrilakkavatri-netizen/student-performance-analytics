```python
import pandas as pd
import numpy as np

# Student details
data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115, 116, 117, 118, 119, 120],

    "Name": ["Anu", "Reethu", "Soni", "Keerthi", "Sai",
             "Harika", "Varun", "Pooja", "Nikhil", "Swathi",
             "Kavya", "Rohit", "Meena", "Arjun", "Divya",
             "Kiran", "Sneha", "Vijay", "Lakshmi", "Rakesh"],

    "Python": [78, 65, 88, 72, 91, 56, 84, 69, 76, 82,
               61, 95, 73, 87, 68, 79, 92, 64, 85, 71],

    "Maths": [74, 70, 91, 68, 88, 59, 79, 73, 81, 85,
              65, 93, 76, 89, 62, 77, 90, 66, 83, 72],

    "English": [82, 68, 85, 75, 90, 63, 81, 71, 78, 88,
                67, 89, 74, 84, 70, 80, 94, 69, 86, 73],

    "Attendance": [88, 76, 91, 82, 95, 68, 87, 79, 85, 92,
                   72, 96, 81, 89, 74, 86, 93, 70, 90, 78]
}

# Create DataFrame
df = pd.DataFrame(data)

# Subjects
subjects = ["Python", "Maths", "English"]

# Total marks
df["Total"] = df[subjects].sum(axis=1)

# Average marks
df["Average"] = np.mean(df[subjects], axis=1)

# Percentage
df["Percentage"] = (df["Total"] / 300) * 100

# Performance
def performance(percentage):
    if percentage >= 85:
        return "Excellent"
    elif percentage >= 70:
        return "Good"
    elif percentage >= 50:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Percentage"].apply(performance)

# Display student report
print("\n===== STUDENT PERFORMANCE REPORT =====\n")
print(df.to_string(index=False))

# Class analysis
print("\n===== CLASS ANALYSIS =====")

print("Number of Students:", len(df))
print("Class Average:", round(np.mean(df["Percentage"]), 2), "%")
print("Highest Percentage:", round(np.max(df["Percentage"]), 2), "%")
print("Lowest Percentage:", round(np.min(df["Percentage"]), 2), "%")

# Top student
top_index = np.argmax(df["Percentage"])

print("\n===== TOP STUDENT =====")
print("Name:", df.loc[top_index, "Name"])
print("Percentage:", round(df.loc[top_index, "Percentage"], 2), "%")

# Subject averages
print("\n===== SUBJECT AVERAGES =====")

for subject in subjects:
    print(subject, ":", round(np.mean(df[subject]), 2))

# Attendance analysis
print("\n===== ATTENDANCE ANALYSIS =====")

low_attendance = df[df["Attendance"] < 75]

if len(low_attendance) > 0:
    print("Students with attendance below 75%:")
    print(low_attendance[["Student_ID", "Name", "Attendance"]].to_string(index=False))
else:
    print("No students have attendance below 75%.")

# Performance count
print("\n===== PERFORMANCE SUMMARY =====")
print(df["Performance"].value_counts())

# Save report
df.to_csv("student_performance_report.csv", index=False)

print("\nReport saved successfully!")
```
